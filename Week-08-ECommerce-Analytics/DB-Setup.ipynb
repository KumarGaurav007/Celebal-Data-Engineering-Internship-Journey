# DB Setup – E-Commerce Order Analytics System
## Introduction

This notebook performs the initial database setup for the E-Commerce Order Analytics System project. The purpose of this notebook is to configure the local environment, create the project database, and verify that the SQLite connection is working correctly.

The raw and cleaned CSV files will be generated and processed in subsequent notebooks.

# Step 1: Import Required Libraries


```python
import sqlite3 
import pandas as pd 
import os
```

# Step 2: Create Required Directories

```python
os.makedirs('data/raw-data', exist_ok=True)
os.makedirs('data/cleaned-data', exist_ok=True)
print('Project directories created successfully.')
```

# Step 3: Create SQLite Database

```python
conn = sqlite3.connect('ecommerce.db')
print('Database created successfully.')
```

# Step 4: Verify Database Connection

```python
cursor = conn.cursor() 
cursor.execute('SELECT sqlite_version();') 
version = cursor.fetchone() 
print('SQLite Version:', version[0])
```

# Step 5: Verify Database File Exists

```python
print(os.path.exists('ecommerce.db'))
```

# Step 6: Close the Connection

```python
conn.close() 
print('Database connection closed.')
```

## Conclusion

In this notebook, the local environment and SQLite database for the E-Commerce Order Analytics System were successfully configured. The project directories were created, the database was initialized, and the connection was verified. This setup provides the foundation for the next phase, where synthetic raw datasets will be generated and stored in the data/raw folder before undergoing data cleaning and analytical processing.

The database is now ready for subsequent data loading and SQL analysis tasks.
