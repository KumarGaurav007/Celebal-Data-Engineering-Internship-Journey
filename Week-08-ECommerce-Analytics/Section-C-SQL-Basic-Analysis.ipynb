# Section-C: SQL Basic Analysis

## E-Commerce Order Analytics System

This notebook performs basic business analysis using SQL queries on the cleaned e-commerce dataset stored in SQLite.

The objective is to extract meaningful insights related to revenue, customers, orders, products, and returns using SQL aggregation and filtering techniques.

## Import Libraries

```python
import pandas as pd
import sqlite3
```

## Connect Database

```python
conn = sqlite3.connect("ecommerce.db")
cursor = conn.cursor()

print("Connected Successfully")
```

## Load Cleaned CSVs into SQLite

```python
customers = pd.read_csv("data/cleaned-data/customers_clean.csv")
products = pd.read_csv("data/cleaned-data/products_clean.csv")
orders = pd.read_csv("data/cleaned-data/orders_clean.csv")
order_items = pd.read_csv("data/cleaned-data/order_items_clean.csv")

customers.to_sql("customers", conn, if_exists="replace", index=False)
products.to_sql("products", conn, if_exists="replace", index=False)
orders.to_sql("orders", conn, if_exists="replace", index=False)
order_items.to_sql("order_items", conn, if_exists="replace", index=False)

print("All tables loaded successfully")
```

## Query 1: Total Revenue per Category

This query calculates total revenue generated by each product category.
Revenue is calculated as:

Revenue = Quantity × Unit Price

```python
query = """
SELECT
    p.category,
    ROUND(SUM(oi.quantity * oi.unit_price),2) AS total_revenue
FROM order_items oi
JOIN products p
ON oi.product_id = p.product_id
GROUP BY p.category
ORDER BY total_revenue DESC;
"""

pd.read_sql(query, conn)
```

## Query 2: Top 10 Customers

This query identifies the top 10 customers based on their total spending.

```python
query = """
SELECT
    c.customer_name,
    ROUND(SUM(oi.quantity * oi.unit_price),2) AS revenue
FROM customers c
JOIN orders o
ON c.customer_id = o.customer_id
JOIN order_items oi
ON o.order_id = oi.order_id
GROUP BY c.customer_name
ORDER BY revenue DESC
LIMIT 10;
"""

pd.read_sql(query, conn)
```

## Query 3: Monthly Order Count

This query shows how many orders were placed each month.

```python
query = """
SELECT
    strftime('%Y-%m', order_date) AS month,
    COUNT(*) AS total_orders
FROM orders
GROUP BY month
ORDER BY month;
"""

pd.read_sql(query, conn)
```

## Query 4: Orders by Status

This query shows the distribution of orders across different statuses.

```python
query = """
SELECT
    status,
    COUNT(*) AS total_orders
FROM orders
GROUP BY status
ORDER BY total_orders DESC;
"""

pd.read_sql(query, conn)
```

## Query 5: Top Selling Products

This query identifies products with the highest quantity sold.

```python
query = """
SELECT
    p.product_name,
    SUM(oi.quantity) AS total_quantity
FROM order_items oi
JOIN products p
ON oi.product_id = p.product_id
GROUP BY p.product_name
ORDER BY total_quantity DESC
LIMIT 10;
"""

pd.read_sql(query, conn)
```

## Query 6: Revenue by Region

This query calculates total revenue generated from each region.

```python
query = """
SELECT
    o.region_code,
    ROUND(SUM(oi.quantity * oi.unit_price),2) AS revenue
FROM orders o
JOIN order_items oi
ON o.order_id = oi.order_id
GROUP BY o.region_code
ORDER BY revenue DESC;
"""

pd.read_sql(query, conn)
```
