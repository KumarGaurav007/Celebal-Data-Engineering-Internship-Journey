# Section-F: Edge Cases and Testing

## E-Commerce Order Analytics System

This notebook validates the robustness of the data pipeline by testing various edge cases and data quality scenarios.

The objective is to identify anomalies, invalid records, and unusual business situations that may affect analytics and reporting accuracy.

## Imports & Connection

```python
import pandas as pd
import sqlite3

conn = sqlite3.connect("ecommerce.db")

print("Database Connected Successfully")
```

## Test Case 1: Orders with Missing Customer IDs

This test identifies orders that do not have a valid customer ID.

```python
query = """
SELECT *
FROM orders
WHERE customer_id IS NULL
OR customer_id = '';
"""

pd.read_sql(query, conn)
```

## Test Case 2: Invalid Discount Percentage

This test checks for discounts greater than 100%.

```python
query = """
SELECT *
FROM order_items
WHERE discount_percent > 100;
"""

pd.read_sql(query, conn)
```

## Test Case 3: Invalid Quantity Values

This test identifies records with zero or negative quantities.

```python
query = """
SELECT *
FROM order_items
WHERE quantity <= 0;
"""

pd.read_sql(query, conn)
```

## Test Case 4: Future Order Dates

This test checks whether any orders contain dates in the future.

```python
query = """
SELECT *
FROM orders
WHERE order_date > DATE('now');
"""

pd.read_sql(query, conn)
```

## Test Case 5: Referential Integrity Check

This test identifies order items whose order_id does not exist in the orders table.

```python
query = """
SELECT oi.*
FROM order_items oi
LEFT JOIN orders o
ON oi.order_id = o.order_id
WHERE o.order_id IS NULL;
"""

pd.read_sql(query, conn)
```

## Test Case 6: Invalid Email Validation

This test identifies customer records containing invalid email formats.

```python
query = """
SELECT *
FROM customers
WHERE email NOT LIKE '%@%';
"""

pd.read_sql(query, conn)
```

## Test Case 7: Duplicate Customer Detection

This test identifies duplicate customer records.

```python
query = """
SELECT
customer_name,
COUNT(*) AS occurrences
FROM customers
GROUP BY customer_name
HAVING COUNT(*) > 1;
"""

pd.read_sql(query, conn)
```

## Data Quality Summary

The following tests were performed:

- Missing customer IDs
- Invalid discount values
- Negative or zero quantities
- Future order dates
- Referential integrity validation
- Email validation
- Duplicate customer detection

These tests help ensure the reliability and accuracy of downstream analytics.

```python
tables = [
    "customers",
    "products",
    "orders",
    "order_items"
]

for table in tables:
    query = f"SELECT COUNT(*) AS total_records FROM {table}"
    count = pd.read_sql(query, conn)

    print(
        f"{table} :",
        count.iloc[0,0],
        "records"
    )
```
