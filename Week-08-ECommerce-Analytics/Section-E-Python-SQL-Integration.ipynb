# Section-E: Python and SQL Integration

## E-Commerce Order Analytics System

This notebook demonstrates the integration of Python and SQL for generating business reports from the e-commerce database.

The objective is to execute SQL queries using Python, retrieve analytical insights, and present them in a structured reporting format.

## Import Libraries

```python
import pandas as pd
import sqlite3
```

##  Database Connection

```python
conn = sqlite3.connect("ecommerce.db")

print("Database Connected Successfully")
```

## Report 1: Total Revenue

This report calculates the overall revenue generated from all orders.

```python
query = """
SELECT
ROUND(SUM(quantity * unit_price),2) AS total_revenue
FROM order_items;
"""

revenue = pd.read_sql(query, conn)

print("===== TOTAL REVENUE REPORT =====")
display(revenue)
```

## Report 2: Top 5 Customers

This report identifies customers generating the highest revenue.

```python
query = """
SELECT
c.customer_name,
ROUND(SUM(oi.quantity * oi.unit_price),2) AS revenue
FROM customers c
JOIN orders o
ON c.customer_id = o.customer_id
JOIN order_items oi
ON o.order_id = oi.order_id
GROUP BY c.customer_name
ORDER BY revenue DESC
LIMIT 5;
"""

top_customers = pd.read_sql(query, conn)

print("===== TOP 5 CUSTOMERS =====")
display(top_customers)
```

## Report 3: Best Selling Products

This report identifies products with the highest quantity sold.

```python
query = """
SELECT
p.product_name,
SUM(oi.quantity) AS quantity_sold
FROM order_items oi
JOIN products p
ON oi.product_id = p.product_id
GROUP BY p.product_name
ORDER BY quantity_sold DESC
LIMIT 5;
"""

products_report = pd.read_sql(query, conn)

print("===== TOP 5 PRODUCTS =====")
display(products_report)
```

## Report 4: Revenue by Region

This report analyzes revenue contribution from each region.

```python
query = """
SELECT
o.region_code,
ROUND(SUM(oi.quantity * oi.unit_price),2) AS revenue
FROM orders o
JOIN order_items oi
ON o.order_id = oi.order_id
GROUP BY o.region_code
ORDER BY revenue DESC;
"""

region_report = pd.read_sql(query, conn)

print("===== REVENUE BY REGION =====")
display(region_report)
```

## Report 5: Order Status Summary

This report shows the number of orders under each order status.

```python
query = """
SELECT
status,
COUNT(*) AS total_orders
FROM orders
GROUP BY status;
"""

status_report = pd.read_sql(query, conn)

print("===== ORDER STATUS REPORT =====")
display(status_report)
```

## Final Business Dashboard

This section combines multiple metrics into a single executive summary report.

```python
total_customers = pd.read_sql(
    "SELECT COUNT(*) AS customers FROM customers",
    conn
)

total_orders = pd.read_sql(
    "SELECT COUNT(*) AS orders_count FROM orders",
    conn
)

total_products = pd.read_sql(
    "SELECT COUNT(*) AS products_count FROM products",
    conn
)

total_revenue = pd.read_sql(
    """
    SELECT ROUND(
    SUM(quantity * unit_price),2
    ) AS revenue
    FROM order_items
    """,
    conn
)

print("========== BUSINESS DASHBOARD ==========")

print(
    "Total Customers :",
    total_customers.iloc[0,0]
)

print(
    "Total Orders :",
    total_orders.iloc[0,0]
)

print(
    "Total Products :",
    total_products.iloc[0,0]
)

print(
    "Total Revenue : ₹",
    total_revenue.iloc[0,0]
)
```
