## Importing the Dataset into MySQL

### Objective

The Superstore dataset is imported into a MySQL database using **Pandas** and **SQLAlchemy**. This approach automates the data loading process, making the workflow reproducible and eliminating the need for manual CSV imports through database tools.

---

### Steps Performed

1. Load the Superstore CSV file into a Pandas DataFrame.
2. Connect to the MySQL database using SQLAlchemy.
3. Import the DataFrame into a table named **`superstore_raw`**.
4. Replace the existing table (if present) to ensure the latest dataset is available for analysis.

---

### Python Libraries Used

- **Pandas** – Reads the CSV file into a DataFrame.
- **SQLAlchemy** – Creates a connection between Python and MySQL.
- **PyMySQL** – Acts as the MySQL database driver for SQLAlchemy.

---


````python
import pandas as pd
from sqlalchemy import create_engine

df = pd.read_csv(
    "data/Superstore.csv",
    encoding="latin1"
)

engine = create_engine(
    "mysql+pymysql://root:1234@localhost/subqueries_sql"
)

df.to_sql(
    "superstore_raw",
    engine,
    if_exists="replace",
    index=False
)
````

### Steeping into the Database

````python
%%sql
USE subqueries_sql;
````

### Tables Created

The database contains the following tables:

1. **customers**
   - Stores customer information.
   - Primary Key: `customer_id`
   - From : `superstore_raw` table, the following columns are extracted: `customer_id`, `customer_name`, `segment`, `city`, `state`, `country`, and `postal_code`.

````python
%%sql
CREATE TABLE customers AS
SELECT DISTINCT
    `Customer ID`,
    `Customer Name`,
    Segment,
    Country,
    City,
    State,
    `Postal Code`,
    Region
FROM superstore_raw;
````

2. **orders**
   - Stores order information.
   - Primary Key: `order_id`
   - Foreign Key: `customer_id`
   - From : `superstore_raw` table, the following columns are extracted: `order_id`, `customer_id`, `order_date`, `ship_date`, and `ship_mode`.

````python
%%sql
CREATE TABLE orders AS
SELECT DISTINCT
    `Order ID`,
    `Customer ID`,
    `Order Date`,
    `Ship Date`,
    `Ship Mode`
FROM superstore_raw;
````

3. **products**
   - Stores product details.
   - Primary Key: `product_id`
   - From : `superstore_raw` table, the following columns are extracted: `product_id`, `product_name`, `category`, and `sub_category`.

````python
%%sql
CREATE TABLE products AS
SELECT DISTINCT
    `Product ID`,
    `Order ID`,
    Category,
    `Sub-Category`,
    `Product Name`,
    Sales,
    Quantity,
    Discount,
    Profit
FROM superstore_raw;
````
