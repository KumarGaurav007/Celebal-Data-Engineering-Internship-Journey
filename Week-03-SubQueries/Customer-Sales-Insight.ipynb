## Business Question 1: Who are the Top 5 Customers?

### Objective

The objective of this analysis is to identify the **top five customers** who have generated the highest total sales.

Identifying top-performing customers helps businesses understand which customers contribute the most revenue and supports strategic decisions related to customer retention, loyalty programs, and personalized marketing.

---

### SQL Query

```python
%%sql
WITH customer_sales AS
(
    SELECT
        `Customer ID`,
        `Customer Name`,
        SUM(Sales) AS total_sales
    FROM superstore_raw
    GROUP BY
        `Customer ID`,
        `Customer Name`
)

SELECT
    `Customer ID`,
    `Customer Name`,
    total_sales
FROM customer_sales
ORDER BY total_sales DESC
LIMIT 5;
```

### Explanation

This query performs the following steps:

1. Calculates the total sales for each customer using the `SUM()` aggregate function.
2. Groups all transactions belonging to the same customer.
3. Sorts customers in descending order based on total sales.
4. Returns only the top five customers using the `LIMIT` clause.


## Business Question 2: Who are the Bottom 5 Customers by Total Sales?

### Objective

The objective of this analysis is to identify the **bottom five customers** who have generated the lowest total sales.

Recognizing low-value customers helps businesses understand purchasing patterns, identify customers with low engagement, and develop strategies to improve customer retention and increase sales.

---

### SQL Query

```python
%%sql
WITH customer_sales AS
(
    SELECT
        `Customer ID`,
        `Customer Name`,
        SUM(Sales) AS total_sales
    FROM superstore_raw
    GROUP BY
        `Customer ID`,
        `Customer Name`
)

SELECT
    `Customer ID`,
    `Customer Name`,
    total_sales
FROM customer_sales
ORDER BY total_sales ASC
LIMIT 5;
```

### Explanation

This query performs the following steps:

1. Calculates the total sales for each customer using the `SUM()` aggregate function.
2. Groups all transactions by customer.
3. Sorts customers in ascending order of total sales.
4. Retrieves the five customers with the lowest total sales.

> **Note:** A customer with low total sales is not necessarily an unprofitable customer. They may be a new customer, an occasional buyer, or someone with the potential to generate higher future revenue through effective engagement strategies.

## Business Question 3: Which Customers Made Only One Order?

### Objective

The objective of this analysis is to identify customers who have placed **only one order** in the Superstore dataset.

Identifying one-time customers helps businesses understand customer retention, purchasing behavior, and potential opportunities for re-engagement through targeted marketing campaigns.

---

### SQL Query

```python
%%sql
SELECT
    `Customer ID`,
    `Customer Name`,
    COUNT(DISTINCT `Order ID`) AS total_orders
FROM superstore_raw
GROUP BY
    `Customer ID`,
    `Customer Name`
HAVING COUNT(DISTINCT `Order ID`) = 1
ORDER BY `Customer Name`;
```

### Explanation

This query performs the following steps:

1. Groups the dataset by customer.
2. Counts the number of unique orders placed by each customer using `COUNT(DISTINCT Order ID)`.
3. Filters the results to include only customers who have placed exactly one order.
4. Sorts the final output alphabetically by customer name.

The `HAVING` clause is used because filtering is performed on an aggregated value.


## Business Question 4: Which Customers Have Above-Average Sales?

### Objective

The objective of this analysis is to identify customers whose **total sales are higher than the average total sales** of all customers.

This analysis helps businesses recognize high-performing customers who contribute significantly to overall revenue and can be targeted for customer retention and loyalty initiatives.

---

### SQL Query

```python
%%sql
WITH customer_sales AS
(
    SELECT
        `Customer ID`,
        `Customer Name`,
        SUM(Sales) AS total_sales
    FROM superstore_raw
    GROUP BY
        `Customer ID`,
        `Customer Name`
)

SELECT
    `Customer ID`,
    `Customer Name`,
    total_sales
FROM customer_sales
WHERE total_sales >
(
    SELECT AVG(total_sales)
    FROM customer_sales
)
ORDER BY total_sales DESC;
```

### Explanation

This query is executed in two stages:

1. **CTE (`customer_sales`)**
   - Groups all transactions by customer.
   - Calculates the total sales for each customer using the `SUM()` aggregate function.

2. **Main Query**
   - Calculates the average of all customers' total sales using a subquery.
   - Filters only those customers whose total sales are greater than the calculated average.
   - Displays the results in descending order of total sales.


## Business Question 5: What is the Highest Order Value per Customer?

### Objective

The objective of this analysis is to determine the **highest order value** placed by each customer.

This helps identify each customer's largest purchase, providing insights into spending behavior and enabling businesses to recognize customers capable of making high-value transactions.

---

### SQL Query

```python
%%sql
SELECT
    `Customer ID`,
    `Customer Name`,
    MAX(Sales) AS highest_order_value
FROM superstore_raw
GROUP BY
    `Customer ID`,
    `Customer Name`
ORDER BY highest_order_value DESC;
```

### Explanation

This query performs the following steps:

1. Groups all transactions by customer.
2. Uses the `MAX()` aggregate function to identify the highest sales value for each customer.
3. Returns each customer's maximum order value.
4. Sorts the results in descending order to display customers with the highest order values first.
