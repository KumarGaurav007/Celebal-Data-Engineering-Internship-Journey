## Q1. What are the key limitations of traditional MapReduce that make Spark a preferred choice for modern big data processing?

### Objective

The objective of this question is to understand the limitations of the traditional **MapReduce** processing model and explain why **Apache Spark** has become the preferred framework for modern big data analytics and data engineering.

---

### Explanation

Apache Hadoop MapReduce was one of the earliest distributed data processing frameworks. Although it can process massive datasets, it has several limitations that reduce its efficiency for modern data processing workloads.

Apache Spark was introduced to overcome these limitations by providing **in-memory processing**, faster execution, and support for multiple workloads such as SQL, machine learning, streaming, and graph processing.

The major limitations of MapReduce are:

| MapReduce Limitation | How Spark Solves It |
|----------------------|---------------------|
| Stores intermediate results on disk after every stage | Stores intermediate data in memory (RAM), significantly improving performance |
| Slow execution for iterative algorithms | In-memory computation makes iterative processing much faster |
| Complex programming model using separate Map and Reduce phases | Provides simple APIs in Python, Scala, Java, and R |
| Limited support for real-time processing | Supports batch, streaming, and interactive analytics |
| Requires multiple MapReduce jobs for complex workflows | Uses Directed Acyclic Graph (DAG) execution to optimize multiple transformations |
| High disk I/O overhead | Reduces unnecessary disk reads and writes |

---

### Comparison

| Feature | Hadoop MapReduce | Apache Spark |
|----------|------------------|--------------|
| Processing | Disk-Based | In-Memory |
| Speed | Slower | Faster |
| Ease of Development | Complex | Simple APIs |
| Machine Learning | Limited | Built-in MLlib |
| Streaming Support | Not Native | Spark Streaming |
| Interactive Queries | Not Supported | Spark SQL |
| Workflow Optimization | Multiple Jobs | DAG Optimization |

---

### Real-World Example

Consider a machine learning algorithm that needs to process the same dataset multiple times.

- **MapReduce** reads the data from disk and writes intermediate results back to disk after every iteration, resulting in significant disk I/O overhead.
- **Spark** keeps the dataset in memory, allowing subsequent iterations to reuse the cached data without repeated disk access.

As a result, Spark can execute iterative algorithms significantly faster than MapReduce.

---

### Practical Significance

Because of its speed, scalability, and rich ecosystem, Apache Spark is widely used for:

- Data Engineering
- ETL Pipelines
- Data Analytics
- Machine Learning
- Real-Time Data Processing
- Business Intelligence

Many modern cloud platforms such as **Azure Databricks**, **AWS EMR**, and **Google Dataproc** use Apache Spark as their primary distributed processing engine.

---

### Conclusion

Traditional MapReduce is reliable for large-scale batch processing but suffers from high disk I/O overhead, slower execution, and a complex programming model. Apache Spark addresses these limitations through in-memory computing, DAG-based execution, and a unified analytics engine, making it the preferred choice for modern big data processing and data engineering applications.

## Q2. Explain how Spark uses In-Memory Computing to speed up iterative machine learning algorithms compared to disk-based systems.

### Objective

The objective of this question is to understand how **Apache Spark's In-Memory Computing** improves the performance of iterative machine learning algorithms compared to traditional disk-based processing systems such as Hadoop MapReduce.

---

### Explanation

Apache Spark performs computations by storing intermediate data in **memory (RAM)** instead of repeatedly writing and reading data from disk.

In many machine learning algorithms, the same dataset is processed multiple times over several iterations. Spark caches this data in memory, allowing it to be reused efficiently without repeated disk access.

Traditional disk-based systems, such as Hadoop MapReduce, store intermediate results on disk after each processing stage. This results in significant disk I/O overhead, increasing execution time.

By minimizing disk operations, Spark provides much faster processing, especially for iterative workloads.

---

### Comparison

| Disk-Based Systems (MapReduce) | Apache Spark (In-Memory Computing) |
|--------------------------------|------------------------------------|
| Reads data from disk in every iteration | Loads data once and keeps it in memory |
| Writes intermediate results to disk | Stores intermediate results in RAM |
| High disk I/O overhead | Minimal disk I/O |
| Slower execution | Faster execution |
| Suitable for batch processing | Suitable for iterative and real-time processing |

---

### Working of In-Memory Computing

The execution flow in Spark is as follows:

```
Load Dataset
      │
      ▼
Store in Memory (RAM)
      │
      ▼
Iteration 1
      │
      ▼
Iteration 2
      │
      ▼
Iteration 3
      │
      ▼
...
      │
      ▼
Final Result
```

Unlike disk-based systems, the dataset remains in memory throughout the iterations, eliminating repeated disk reads and writes.

---

### Real-World Example

Consider training a Machine Learning model using **Logistic Regression**.

The algorithm repeatedly updates its model parameters by processing the same training dataset over multiple iterations.

- **MapReduce**
  - Reads the dataset from disk in every iteration.
  - Writes intermediate results back to disk.
  - Results in slower execution due to frequent disk I/O.

- **Spark**
  - Loads the dataset into memory once.
  - Reuses the cached dataset during every iteration.
  - Significantly reduces execution time and improves performance.

---

### Practical Significance

Spark's in-memory computing is especially beneficial for:

- Machine Learning
- Data Science
- Graph Processing
- Interactive Data Analysis
- Real-Time Analytics
- Large-scale ETL Pipelines

This capability makes Spark one of the most widely used frameworks in modern Data Engineering and Big Data applications.

---

### Conclusion

Apache Spark accelerates iterative machine learning algorithms by keeping data in memory throughout the computation process. Unlike traditional disk-based systems, Spark minimizes disk I/O, enabling faster execution, better resource utilization, and improved performance for large-scale data processing tasks.

## Q4. Given a DataFrame `df_sales`, write a query to filter for rows where the region is **'West'** and then group by **product_category** to find the average **sale_amount**.

````python
from pyspark.sql.functions import count

result = (
    df
    .groupBy("city")
    .agg(count("*").alias("total_records"))
    .filter("total_records > 100")
    .orderBy("total_records", ascending=False)
)

result.show()
````

## Q7. How does the immutability of Spark DataFrames affect how you perform "data cleaning" steps like dropping columns or renaming them?

### Objective

The objective of this question is to understand the concept of **DataFrame immutability** in Apache Spark and how it impacts common data cleaning operations such as dropping columns, renaming columns, or modifying data.

---

### Explanation

In Apache Spark, **DataFrames are immutable**, which means that once a DataFrame is created, its contents cannot be modified directly.

Instead of changing the existing DataFrame, every transformation creates and returns a **new DataFrame** while leaving the original DataFrame unchanged.

This design enables Spark to optimize execution, improve fault tolerance, and support distributed processing across multiple nodes.

---

### Example

**Dropping a Column**

```python
# Create a new DataFrame without the status column
df_clean = df.drop("status")

df_clean.show()
```

**Renaming a Column**

```python
# Rename sale_amount to total_sales
df_updated = df.withColumnRenamed(
    "sale_amount",
    "total_sales"
)

df_updated.show()
```

In both examples:

- The original DataFrame (`df`) remains unchanged.
- A new DataFrame (`df_clean` or `df_updated`) is created with the requested modifications.

---

### Processing Workflow

```
Original DataFrame
        │
        ▼
Apply Transformation
(drop / rename / filter)
        │
        ▼
Create New DataFrame
        │
        ▼
Original DataFrame
Remains Unchanged
```

---

### Why Spark Uses Immutability

Immutability provides several advantages:

- Improves fault tolerance through lineage.
- Prevents accidental modification of data.
- Enables parallel execution across multiple machines.
- Allows Spark to optimize execution using its Catalyst Optimizer.
- Makes DataFrame transformations more predictable and reliable.

---

### Business Significance

In large-scale ETL pipelines, multiple transformations are applied to raw data before it is loaded into a data warehouse.

Since every transformation creates a new DataFrame:

- Raw data remains preserved.
- Intermediate processing steps can be reproduced.
- Pipelines become easier to debug and maintain.
- Data quality improves by avoiding unintended changes.

---

### Conclusion

Spark DataFrames are immutable, meaning they cannot be modified after creation. Every data cleaning operation—such as dropping columns, renaming columns, filtering records, or handling null values—returns a new DataFrame while preserving the original dataset. This immutability enhances reliability, scalability, and fault tolerance, making Spark highly suitable for distributed data processing.

## Q11. Explain the **"Shuffle"** process that occurs during a grouping operation. Why is it considered a **Wide Transformation**?

### Objective

The objective of this question is to understand the **Shuffle** process in Apache Spark and explain why operations such as **groupBy()** are classified as **Wide Transformations**.

Understanding shuffle is important because it directly impacts the performance of Spark applications.

---

### Explanation

A **Shuffle** is the process of **redistributing data across different partitions or worker nodes** during the execution of certain Spark operations.

When an operation such as **groupBy()**, **join()**, or **orderBy()** is executed, Spark must move data between partitions so that records with the same key are placed together.

This movement of data across the cluster is called a **Shuffle**.

Since network communication and disk I/O are involved, shuffle is considered one of the most expensive operations in Spark.

---

### Why Does Shuffle Occur?

Suppose a DataFrame is distributed across multiple partitions.

```
Partition 1

Delhi
Mumbai
Delhi

------------------

Partition 2

Pune
Mumbai
Delhi
```

If we execute:

```python
df.groupBy("city").count()
```

Spark cannot calculate the final count directly because records for the same city exist in different partitions.

Therefore, Spark redistributes the records so that all rows belonging to the same city are placed together before performing the aggregation.

---

### After Shuffle

```
Partition A

Delhi
Delhi
Delhi

------------------

Partition B

Mumbai
Mumbai

------------------

Partition C

Pune
```

Now Spark can easily calculate:

```
Delhi → 3

Mumbai → 2

Pune → 1
```

---

### Why is Shuffle a Wide Transformation?

Spark transformations are classified into two types:

| Narrow Transformation | Wide Transformation |
|-----------------------|--------------------|
| Data remains within the same partition. | Data is moved between multiple partitions. |
| No network communication. | Requires network communication (shuffle). |
| Faster execution. | Slower execution due to data movement. |

Operations such as:

- groupBy()
- join()
- distinct()
- repartition()
- orderBy()

are **Wide Transformations** because they require Spark to redistribute data across partitions.

---

### Processing Workflow

```
Original Data

Partition 1
Partition 2
Partition 3

        │

        ▼

groupBy()

        │

        ▼

Shuffle

(Data Redistributed)

        │

        ▼

Aggregation

        │

        ▼

Final Result
```

---

### Business Significance

Although shuffle enables Spark to perform complex distributed computations, it is one of the most resource-intensive operations.

Excessive shuffling can:

- Increase execution time.
- Consume network bandwidth.
- Increase disk I/O.
- Reduce overall cluster performance.

Therefore, Data Engineers should design Spark jobs to minimize unnecessary shuffle operations whenever possible.

---

### Conclusion

Shuffle is the process of redistributing data across partitions during operations such as **groupBy()**, **join()**, and **orderBy()**. Since data must move between worker nodes, shuffle involves network communication and is therefore classified as a **Wide Transformation**. Efficient Spark applications aim to reduce unnecessary shuffles to improve performance.

When **`inferSchema=true`** is enabled, PySpark automatically examines the data and attempts to determine the appropriate data type for each column.

While this is convenient, it can become problematic when date columns contain inconsistent formats.

For example, consider the following values in a date column:

| Raw Date Values |
|-----------------|
| 2025-01-15 |
| 15/01/2025 |
| Jan-15-2025 |
| NULL |

Since multiple date formats exist in the same column, Spark may not correctly identify the column as a **DateType**. Instead, it may infer the column as a **StringType**, or some values may be interpreted incorrectly.

This can lead to data quality issues and errors during further processing.

---

### Risks of Using `inferSchema=true`

- Incorrect data type inference (e.g., Date interpreted as String).
- Parsing failures for inconsistent date formats.
- Incorrect filtering and sorting of date values.
- Errors in time-based calculations and aggregations.
- Unreliable ETL pipelines due to inconsistent schemas.

---

### Example

```python
df = spark.read.csv(
    "data/pyspark_assignment_dataset_v2.csv",
    header=True,
    inferSchema=True
)
```

If the source data contains inconsistent date formats, Spark may infer the column as:

```
transaction_date : string
```

instead of:

```
transaction_date : date
```

---

### Recommended Approach

Instead of relying entirely on `inferSchema=true`, Data Engineers often:

- Define the schema explicitly using `StructType`.
- Standardize date formats before processing.
- Use functions such as `to_date()` or `to_timestamp()` with a specified format.
- Validate and clean date columns during data ingestion.

Example:

```python
from pyspark.sql.functions import to_date

df_clean = df.withColumn(
    "transaction_date",
    to_date("transaction_date", "yyyy-MM-dd")
)
```

In real-world ETL pipelines, inconsistent date formats are a common data quality issue.

Incorrect schema inference can lead to:

## Q15. Write a final processing pipeline that:

- Filters out duplicate records.
- Fills null prices with `0`.
- Groups by `store_id` to calculate total revenue.

### Objective

The objective of this task is to build a complete data processing pipeline using PySpark. The pipeline demonstrates common data engineering operations, including duplicate removal, missing value handling, and revenue aggregation.

This workflow simulates a typical ETL (Extract, Transform, Load) process where raw transactional data is cleaned and transformed into meaningful business insights.

---

### PySpark Code

````python
from pyspark.sql.functions import sum

# Step 1: Remove duplicate records
df_clean = df.dropDuplicates(["user_id", "transaction_date"])

# Step 2: Fill null values in the price column with 0
df_clean = df_clean.na.fill(
    value=0,
    subset=["price"]
)

# Step 3: Calculate total revenue for each store
result = (
    df_clean
    .groupBy("store_id")
    .agg(
        sum("price").alias("total_revenue")
    )
    .orderBy("total_revenue", ascending=False)
)

result.show()
````

### Explanation

The pipeline consists of three major stages:

#### Step 1 – Remove Duplicate Records

Duplicate transactions are removed based on the combination of:

- `user_id`
- `transaction_date`

This ensures that repeated transactions are not counted multiple times.

---

#### Step 2 – Handle Missing Values

Missing values in the **price** column are replaced with **0** using `.na.fill()`.

This prevents aggregation functions from producing inaccurate or incomplete results.

---

#### Step 3 – Calculate Store Revenue

The cleaned dataset is grouped by **store_id**, and the total revenue for each store is calculated using the `sum()` aggregate function.

Finally, the results are sorted in descending order of revenue.
