## Step 1 — Load a CSV into a DataFrame

```python
import pandas as pd
import numpy as np

df = pd.read_csv("sample.csv", encoding="cp1252")
df.head()
```

## Step 2 — Explore the data

```python
# Number of rows (observations) and columns (features)
df.shape
```

```python
# Last 5 rows — the data can look different at the end than at the top
df.tail()
```

```python
# All column names
df.columns
```

```python
# Data type of each column (int64 = whole numbers, float64 = decimals, object = text)
df.dtypes
```

```python
# One-line summary: size, dtypes, non-null counts, memory usage
df.info()
```

```python
# Numeric summary statistics: count, mean, min, max, quartiles...
df.describe()
```

## Step 3 — Handle missing values

### 3a) Identify missing values

```python
# How many missing values per column?
df.isnull().sum()
```

```python
# What percentage of the table is missing in each column?
df.isnull().sum() / len(df) * 100
```

### 3b) Fill them — the strategy depends on the column

* **`Postal Code`** (an identifier) → fill with the *most common* value (`mode`)
* **`Region`** (categorical text) → fill with the *most common* value (`mode`)
* **`Sales`** (numbers) → fill with the *median* (resistant to outliers)

```python
df["Postal Code"] = df["Postal Code"].fillna(df["Postal Code"].mode()[0])
df["Region"]      = df["Region"].fillna(df["Region"].mode()[0])
df["Sales"]       = df["Sales"].fillna(df["Sales"].median())

# Confirm: every column now has 0 missing values
df.isnull().sum()
```

### 3c) The alternative — drop them

Sometimes filling makes no sense (too much missing, or a value that genuinely can't
be guessed). Then drop the offending rows with `dropna()`. We already filled
everything, so we just preview what that would look like on a copy.

```python
# What WOULD happen if we dropped every row with any missing value?
df.dropna().shape
```

## Step 4 — Basic operations: filter rows & select columns

### 4a) Filter rows with a condition


```python
# Orders over $1,000
big_orders = df[df["Sales"] > 1000]
big_orders.shape
```

```python
# Combine conditions with & (and) — wrap each condition in parentheses
consumers_west = df[(df["Segment"] == "Consumer") & (df["Region"] == "West")]
consumers_west.shape
```

### 4b) Select columns


```python
df[["Order ID", "Product Name", "Sales", "Quantity"]].head()
```

```python
# .loc: rows 0-4, columns "Order ID" and "Sales"
df.loc[0:4, ["Order ID", "Sales"]]
```

```python
# .iloc: rows 0-4, column positions 1 (Order ID) and 17 (Sales)
df.iloc[0:4, [1, 17]]
```

## Step 5 — Remove duplicates


```python
# How many duplicate rows are hiding in the data?
df.duplicated().sum()
```

```python
df = df.drop_duplicates()
print("Rows after removing duplicates:", len(df))
```

## Step 6 — Create a derived column


```python
df["Unit_Price"] = df["Sales"] / df["Quantity"]
df["total_amount"] = df["Unit_Price"] * df["Quantity"]

df[["Sales", "Quantity", "Unit_Price", "total_amount"]].head()
```

```python
# Sanity check: within 1 cent of Sales, for every row?
diff = (df["total_amount"] - df["Sales"]).abs()
print("Largest difference from Sales:", diff.max())
(diff < 0.01).all()
```

## Step 7 — Save the cleaned dataset

```python
df.to_csv("superstore_cleaned.csv", index=False)
print("Saved superstore_cleaned.csv")
```

```python
# Verify the saved file loads back cleanly
cleaned = pd.read_csv("superstore_cleaned.csv")
print("Loaded back:", cleaned.shape)
print("Missing values:", cleaned.isnull().sum().sum())
print("Duplicate rows:", cleaned.duplicated().sum())
```
